### Pseudocode

-   Write pseudocode to check if given $n \in \mathbb{N}$ is
    [primenumber](https://en.wikipedia.org/wiki/Prime_number).

``` text
ask n

i=n-1

loop:
  if n/i is integer
    return False
  
  i=i-1
  if i=1
    return True
goto loop
```

### Alternative syntax

``` text
n <- user input

i <- n-1
Loop
  If n/i is integer
    Return False
  EndIf
  
  i <- i-1
  If i == 1
    Return True
  EndIf
EndLoop
```

### Pseudocode, version 2

-   Write pseudocode to check if given $n \in \mathbb{N}$ is prime.

``` text
ask n

for i=2..(n-1)
  if n/i is integer
    return False

return True
```

### Pseudocode, version 3

-   Write pseudocode to check if given $n \in \mathbb{N}$ is prime.

``` text
n <- user input

If exists integer i , 1<i<n, s.t n is divisible by i
  Return False
EndIf

Return True
```

### Reading a flowchart

-   Start from a start node. There shall be only one.

-   Follow arrows from one node to another. Only one arrow shall be
    chosen at each step.

-   Only decision node shall have more than one outgoing arrows.

-   Follow instructions inside the node.

-   If end node is reached, then the algorithm is done. There shall be
    none outgoing arrows at the node.

-   There can be multiple end nodes, each of them representing separate
    end state.

### Why comments?

-   At beginning of a file:

    -   Who, and when, have created the file

    -   Who, when, and what, have modified the file

    -   Purpose of the file

-   Before function/method:

    -   What function does

    -   Function parameters

    -   Return value

-   [Best practices for writing code
    comments](https://stackoverflow.blog/2021/12/23/best-practices-for-writing-code-comments/)
