### `list` properties and indexing

-   `l` is a `list`

   

| `len(l)`           | Number of items in list (length)            |
|:-------------------|:--------------------------------------------|
| `l.append(n)`      | Append `n` to end of the `list`             |
| `l.sort()`         | Sort the `list` – Items must be comparable  |
| `min(l)`, `max(l)` | Find smallest/largest element on the `list` |
|                    | – Items must be comparable                  |
| `l[0]`             | The 1st item in list                        |
| `l[1]`             | The 2nd item in list                        |
| `l[n]`             | The (n+1)th item in list                    |
| `l[-1]`            | The last item in list                       |
| `l[-2]`            | The 2nd last item in list                   |
| `l[1:]`            | From 2nd item to end of the list            |
| `l[n:]`            | From (n+1)th item to end of the list        |
| `l[1:3]`           | From 2nd item to 3rd item of the list       |
| `l[n:m]`           | From (n+1)th item to mth item of the list   |

### `list.copy()`

-   `list` variable is reference to the `list`

-   `another_list=my_list` creates new reference variable

-   Both variables sees the same list

-   To create a copy of the `list`, use `list.copy()`

-   Use `list.copy()` only when required, it can slow down the code!

### Example

Run codes one after another in order!

````python
%reset -f

my_list=[1, 2, 3]
my_list
````

````python
another_list=my_list
another_list
````

## `for` + `list`

### `for`

-   `for`-loop "range" can be replaced by a `list`

-   A loop variable `var` takes value from an `iterable`

``` PseudoPython
for var in iterable:
   statement
```

-   `iterable` is a collection of objects – for example, a `list`

-   `range`-function returns an iterable

### Example

````python
%reset -f

# Source file: src/lesson_09_10/examplelist01.py

for i in range(5):
    print(i, end=' ')
print()
````

````python
for i in [0, 1, 2, 3, 4]:
    print(i, end=' ')
print()
````

````python
numbers=[0, 1, 2, 3, 4]
for i in numbers:
    print(i, end=' ')
print()
````

-   `type` returns type information of an object

### `enumerate`

-   [`enumerate`](https://www.w3schools.com/python/ref_func_enumerate.asp)
    enumerates the `list`

    > *iterator*

    items

-   returns index and item

````python
%reset -f

# Source file: src/lesson_09_10/examplelist04.py

my_list=[-1, 2, 'tdtd',  6, -1, 11, 0.1]
````

````python
#Bad code!
for i in range(len(my_list)):
    print(f'{i}: {my_list[i]}')


print(24*'-')
````

````python
#Preferred solution uses enumerate
for i, item in enumerate(my_list):
    print(f'{i}: {item}')
````

### `zip`

-   [`zip`](https://www.w3schools.com/python/ref_func_zip.asp) pairs the
    `list`

    > *iterator*

    items

````python
%reset -f

# Source file: src/lesson_09_10/examplelist05.py

first_name=['Bob', 'Alice', 'John']
last_name=['Baker', 'Anderson', 'Smith']
````

````python
#Bad code
for i in range(len(first_name)):
    print(first_name[i], last_name[i])

print(24*'-')
````

````python
#preferred solution
for fn, ln in zip(first_name, last_name):
    print(fn, ln)
````
