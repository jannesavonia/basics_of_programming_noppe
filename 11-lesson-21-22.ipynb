## Object Oriented Programming (OOP) – basics

### Procedural programming

### OOP

-   [Object oriented
    programming](https://en.wikipedia.org/wiki/Object-oriented_programming)

-   Programming paradigm

    > *In practice, most programs are written with mixed programming
    > paradigm*

-   Concept of *object*, containing data and methods (=code/functions)

-   Object methods can manipulate object data or return state of the
    object

-   Object type $\approx$ class

-   Student class

    -   Data: Name, nationality, courses completed, etc.

    -   Methods: Total cr, add new course, etc.

-   Complex number class

    -   Data: real part, imaginary part

    -   Methods: Absolute value, add, multiply, etc.

-   Object $\approx$ instance of the class

-   We have utilized objects, for example `str`, `list`, `dict`, `set`,
    etc.

## OOP – member variables, constructor

### Defining a class

-   `class` keyword

``` PseudoPython
class ClassName:
    statement_1
    .
    .
    .
    statement_N    
```

-   `statement_i` – In a typical case function/variable definition

-   Can be anything else – will be executed on object creation

### Example

````python
%reset -f

# Source file: src/lesson_21_22/exampleclass01.py

class Coordinate:
    def init(self, x, y):
        #If x or y does not exist, it will be created here
        #self.x = This object variable, x = parameter
        self.x=x
        self.y=y
    
    def print(self):
        print(f'{(self.x, self.y)}')
````

````python
if __name__=='__main__':
    point1=Coordinate()
    point2=Coordinate()
````

````python
    point1.init(-1, -2)
    point2.init(3, 4)
````

````python
    point1.print()
    point2.print()
````

### On the example

-   `point1` has it’s own `x` and `y`, and `point2` has it’s own

-   `Coordinate()` returns an `Coordinate`-type object

-   `.`-notation is used to refer object internals

-   `self` is a reference to the object itself

### Constructor

-   We have initialized values of class members by `init`-function after
    object creation

-   Class constructor is a function which is executed automatically once
    the object is created

-   `__init__(parameters)`

-   There can be $0\dots N$ `parameters`

-   Use constructor to initialize class variables!

### Example

````python
%reset -f

# Source file: src/lesson_21_22/exampleclass03.py

class Coordinate:

    #Constructor
    def __init__(self, x, y):
        self.x=x
        self.y=y

    def move(self, dx, dy):
        print(f'move {(dx, dy)}')
        self.x+=dx
        self.y+=dy
        print(f'new position is {(self.x, self.y)}')
        
    def print(self):
        print(f'{(self.x, self.y)}')
````

````python
if __name__=='__main__':
    p=Coordinate(-1, -2)
    p.print()
````

````python
    p.move(2, 4)
    p.print()
````

### Class variables/methods

-   Also known as [static
    variables](https://en.wikipedia.org/wiki/Class_variable)/methods

-   Shared by all instances of the class

-   Declared inside the class, but outside of any instance method

-   `ClassName.`-notation

``` PseudoPython
class ClassName:
    static_var
    def static_method(parameters1):
        statements
        
    def method(self, parameters2):
        ...
        ClassName.static_var=statement
```

### Example

````python
%reset -f

# Source file: src/lesson_21_22/exampleclass02.py

class MyClass:
    shared=10 #Class variable
    
    def print_shared(): #Class method
        print(f'MyClass.shared={MyClass.shared}')

    def __init__(self, i=0):
        self.object_add=0
        #It is possible to call methods from constructor
        self.add(i)

    def add(self, i): #Instance method
        print(f'MyClass.add({i})')
        self.object_add+=i
        #Modifying class variable
        MyClass.shared+=i

    #Converts object to string
    def __str__(self): #Instance method
        return f'{self.object_add}, {MyClass.shared}'
````

````python
MyClass.print_shared()
````

````python
print('Create a')
a=MyClass(3)
print(f'a={a}')
````

````python
a.add(7)
print(f'a={a}')
````

````python
print('Create b')
b=MyClass(3)
print(f'b={b}')
````

````python
MyClass.print_shared()
````

-   `__str__` is [special method
    name](https://diveintopython3.net/special-method-names.html)

    > *`__init__` is, also, a special method.*

-   The name is reserved for a special purpose, and should not be used
    for anythin else
