### Example – `for`

```python
%reset -f

# Source file: src/lesson_11_12/exampledict01.py

student={
    'name'    : 'Janne Koponen',
    'math 1'  : 4,
    'bop'     : 5,
    'ongoing' : ['finnish', 'elements of AI']
}

print(student)
print(24*'-')
```

```python
print('Keys:\n')
for k in student:
    print(k)
print(24*'-')
```

```python
print('Keys:\n')
for k in student.keys():
    print(k)
print(24*'-')
```

```python
print('Values, except math:\n')
for k in student.keys():
    if k!='math 1':
        print(student[k])
print(24*'-')
```

```python
print('Values:\n')
for v in student.values():
    print(v)
print(24*'-')
```

```python
print('Key:Value\n')
for k,v in student.items():
    print(f'{k}: {v}')
print(24*'-')
```

### Example – Modifying `dict`

```python
%reset -f

# Source file: src/lesson_11_12/exampledict02.py

#Create an empty dictiorary
my_dict={}
```

```python
#Key 'first' does not exist
# => Create a new item  
my_dict['first']='the only item'
print(my_dict)
```

```python
#Key 'another' does not exist
# => Cread a new item
my_dict['another']='another item'
print(my_dict)
```

```python
#Key 'first' exists
# => Modify/replace existing value
my_dict['first']='the first item'
print(my_dict)
```

```python
#Remove item with index 'another'
my_dict.pop('another')
print(my_dict)
```

```python
#Update dictionary
my_dict.update({'second': 2, 'third': 3})
print(my_dict)
```

```python
#Remove all elements
my_dict.clear()
print(my_dict)
```

## `set`

### Basics

-   To store collection of **unique items**

    -   Duplicates not allowed

-   Unordered

    -   Items can appear in different order every time you access them

    -   No indexes

-   Item can’t be modified, but

    -   You can remove and add new items

-   Curly brackets

-   Creating an empty `set`:

          my_set=set()  

### Example

```python
%reset -f

# Source file: src/lesson_11_12/exampleset01.py

my_cars={'Opel', 'Toyota', 'BMW'}
print(my_cars)
```

```python
#Add an item
my_cars.add('Porche')
print(my_cars)
```

```python
#Add duplicate
my_cars.add('Opel')
print(my_cars)
```

```python
#Remove 'Porche'
my_cars.remove('Porche')
print(my_cars)
```

```python
for car in my_cars:
    print(car)
```

```python
your_cars={'Mercedes', 'Maserati', 'Opel'}
```

```python
#Join sets
our_cars=my_cars.union(your_cars)
print(our_cars)
```

```python
#Intersection of sets (We both have same car)
both_have=my_cars.intersection(your_cars)
print(both_have)
```
