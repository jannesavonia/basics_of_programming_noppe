## Functions

### Basics

-   A function is a block of code which only runs when it is called.

-   In most cases function has a *name* (Lambda functions does not have
    name!)

-   Function is called by name and arguments surrounded by brackets

-   You can pass data, known as *parameters*, into a function.

-   A function *can return data* as a result.

-   We have already used functions, for example, `len`, `input`, and
    `print`

````python
%reset -f

msg='Hello World!'
len(msg)
````

-   `len` = Function to call

-   `msg` = An argument

-   `12` = Result

### Example

````python
%reset -f

# Source file: src/lesson_13_14/examplefunction01.py

name=input('Give your name: ')

#'input returned' and name are arguments
print('input returned', name)
````

### Defining functions

``` PseudoPython
def function_name(parameters):
   code
```

-   `def` keyword begins defining a function

-   `parameters` contains $0\dots N$ parameters

    -   Parameters are separated by comma

-   `return` keyword terminates function (and returns a result)

    -   Function terminates, also, when it reaches end of `code`

### Example

````python
%reset -f

# Source file: src/lesson_13_14/examplefunction02.py

#Define function say_hello with no parameters
def say_hello():
    print('Hello!')
````

### Example – Multiple `return` points

````python
%reset -f

# Source file: src/lesson_13_14/examplefunction05.py

#Define a function with multiple return points
def are_even(a, b):
    if a%2!=0:
        return False
    if b%2!=0:
        return False

    return True
````

````python
i=int(input('Give an integer: '))
j=int(input('Give another integer: '))
````

````python
if are_even(i, j):
    print(f'{i} and {j} are both even.')
else:
    print(f'At lest one of {i} and {j} is not even.')
````

### Variable visibility

-   Variable is visible only inside the region, *scope*, it is created

-   A variable inside a function is not visible outside the function

-   `global` – Make global variable available

    -   Global variable $\approx$ Variable created inside main program

````python
%reset -f

def set_x(a):
  #x is available inside set_x function
  x=a
````

````python
set_x(10)
````

````python
#x is not available here => Error
print(x)
````

### Example – `global`

````python
%reset -f

# Source file: src/lesson_13_14/examplefunction07.py

def increase_x(y):
    print(f'increase_x({y})')
    #Make x available inside the function
    global x
    x+=y
````

````python
x=0
print('x =', x)
````

````python
increase_x(10)
print('x =', x)
````

### Example – Multiple return values

````python
%reset -f

# Source file: src/lesson_13_14/examplefunction08.py

#Returns the first and the last items on list
def first_and_last(l):
    first=l[0]
    last=l[-1]

    return first, last
````

````python
f,l=first_and_last([1, 2, 'last'])
print(f'f={f}, l={l}')
````

````python
#We need only the last value
_,l=first_and_last([1, 2, 'last'])
print(f'l={l}')
````

````python
#We need only the first value
f,_=first_and_last([1, 2, 'last'])
print(f'f={f}')
````

````python
#Both values in one variable 
both=first_and_last([1, 2, 'last'])
print(f'both={both}, type={type(both)}')
````

### Example – Warning with `list`s

````python
%reset -f

# Source file: src/lesson_13_14/examplefunction09.py

def add_one(x):
    x+=1
    print('x =',x,'in add_one')
````

````python
x=0
print('x =',x, 'in main')
````

````python
add_one(x)
print('x =',x, 'in main\n')
````

````python
#l is reference to list => Modifying
#l modifies, also, the original list
def add_one_list(l):
    for idx in range(len(l)):
        l[idx]+=1
    print('l =',l,'in add_one_list')
````

````python
l=[1, 2, 3]
print('l =',l, 'in main')
````

````python
add_one_list(l)
print('l =',l, 'in main')
````

### Why functions?

1.  Reusability

    -   You can use same function in multiple programs

2.  Make code easier to read and maintain

    -   A function is like a ’mini program’

    -   Hides details of the implementation

    -   Hides complexity

3.  Testing

    -   Once function is well tested, it is safe to use
