## Exceptions

### What are exceptions?

-   Exceptions are mechanism to handle runtime errors

-   For example, user gives string `’one’` when asked an integer

-   Conversion throws an exception

-   If the exception is not handled correct, the software crashes

``` text
In [3]: i=int(input('Give an integer: '))
Give an integer: 34

In [4]: i
Out[4]: 34

In [5]: i=int(input('Give an integer: '))
Give an integer: one
---------------------------------------------------------------------------
ValueError                                Traceback (most recent call last)
Cell In[5], line 1
----> 1 i=int(input('Give an integer: '))

ValueError: invalid literal for int() with base 10: 'one'
```

````python
%reset -f

# Source file: src/lesson_17_18/exampleexception01.py

try:
    i=int(input('Give an integer: '))
    
    print(f'{i} is an integer, great!')
except:
    print('I asked an integer, is it so hard to follow instructions?')
````

### Example – `finally`

````python
%reset -f

# Source file: src/lesson_17_18/exampleexception05.py

def print_add(a, b):
    try:
        print(f'{a}+{b}={a+b}')
    except:
        print('Got an exception!')
    finally:
        print('Now in finally block.\n')


print_add(5, 8)
````

### Example

````python
%reset -f

# Source file: src/lesson_17_18/exampleexception07.py

from math import sqrt

class ValueTooSmallError(Exception):
    pass
````

````python
class ValueTooBigError(Exception):
    def __init__(self, m):
        self.msg=m
````

````python
def my_func(x):
    if x<-1:
        print('raise ValueTooSmallError')
        raise ValueTooSmallError
    if x>7:
        print('raise ValueTooBigError')
        raise ValueTooBigError(f'{x}>7')
    return sqrt(x+1)
````

````python
while True:
    try:
        x=float(input('Give a number -1...7: '))
        res=my_func(x)
        break
    except ValueError:
        print('Got ValueError')
    except ValueTooSmallError:
        print('Got ValueTooSmallError')
    except ValueTooBigError as vtbe:
        print(f'Got ValueTooBigError, msg="{vtbe.msg}"')
    except:
        print('Got an exception')

print(f'Result is {res}')
````

## File I/O

### File

-   So far, when the program has terminated, all information is lost

-   [Files](https://en.wikipedia.org/wiki/Computer_file) are tool to
    save data over the software restarting

-   Files are identified by it’s name

    > *Including path*

-   Typically, files are located on [File
    system](https://en.wikipedia.org/wiki/File_system)

-   File can contain different type of information, for example:

    -   Text

    -   Image

    -   Movie

    -   Computer program

    -   Database

    -   Any combination of these, and much more

-   Files can be read, write and modified

-   Note: Directory separator $\backslash$ works on Windows, but $/$
    works on most operating systes, including Windows

### Using file

-   Most basic operation with files contains

    1.  Open file

        -   Reserve a file to use with this program

        -   Returns handle (file object), which is utilized at following
            steps

        -   File can be opened in read, write, or read+write mode

    2.  Read/Write/Modify file

    3.  Close file

        -   Commit changes

-   These steps can be hidden from the user inside a function

-   Python contains many different ways to work with files

-   In following, we will discus text files

    > *There exists, also, binary files*

### `open`

-   Opens a file

-   Returns file like object – Type depends on parameters

``` PseudoPython
open(file, mode='r')
```

-   `file` = filename

-   `mode` = the mode in which the file is opened

    -   `’r’` = reading

    -   `’w’` = writing, delete existing content

    -   `’a’` = writing, append to end of the file

> *For all features and all parameters, see [the detailed
> documentation](https://docs.python.org/3/library/functions.html#open)*

### `read`

-   Reads data from the file

-   Returns string

``` PseudoPython
f.read(size=-1)
```

-   `size` = number of bytes to read, $-1$=all

-   `f` = file object

> *For more details, see [detailed
> documentation](https://docs.python.org/3/library/io.html#id1)*

### `write`

-   Write data to the file

-   Returns number of bytes written

``` PseudoPython
f.write(s)
```

-   `s` = string to write

-   `f` = file object

> *For more details, see [detailed
> documentation](https://docs.python.org/3/library/io.html#id1)*

### `close`

-   Flush, and closes the file

``` PseudoPython
f.close()
```

-   `f` = file object

> *For more details, see [detailed
> documentation](https://docs.python.org/3/library/io.html#i-o-base-classes)*

### Example

````python
%reset -f

# Source file: src/lesson_17_18/examplefile05.py

datafile='students.txt'
separator=','
course_separator=':'
````

````python
def write_data(students):
    f=open(datafile, 'w')
    for name, studies in students.items():
        f.write(name)
        for course, grade in studies.items():
            f.write(separator)
            f.write(f'{course}{course_separator}{grade}')
        f.write('\n')
    f.close()
````

````python
def read_data():
    f=open(datafile, 'r')
    filedata=f.readlines()
    f.close()

    students={}
    for row in filedata:
        rowdata=row.split(separator)
        name=rowdata[0]
        courses={}
        course_data=rowdata[1:]
        for course in course_data:
            course_name, course_grade=course.split(course_separator)
            courses[course_name]=int(course_grade)
        students[name]=courses

    return students
````

````python
def print_data(students):
    print(24*'-')
    for name, studies in students.items():
        print(name)
        for course, grade in studies.items():
            print(f'    {course} - {grade}')
        print(24*'-')
````
