## Blocks

### Blocks

-   A block is a piece of Python program text that is executed as a unit

-   Block is marked with indentation

-   Block:

    -   Function body

    -   Class definition

    -   etc.

### Example

````python
%reset -f

# Source file: src/lesson_05_06/exampleblock01.py

def my_print(a):
    print(32*'-')
    print(a)
    print(32*'-')


for i in range(10):
    if i%2==0:
        my_print(i)

    if i>5:
        print(i)
````

## `if-elif-else`

### `if`-statement

-   `if`-statement is used for conditional execution

``` PseudoPython
if condition:
   statement
```

-   If `condition` is evaluated as `True`, then `statement` is executed

-   If `condition` is evaluated as `False`, then `statement` is not
    executed

-   `statement` can contain one or many lines

-   Indentation

### Example

````python
%reset -f

# Source file: src/lesson_05_06/exampleif01.py

age=18

if age>=18:
    print('You are adult.')
    print('If you are Finnish citizen,')
    print('then you can vote.')
````

-   Modify `age` to see how `if`-statement works.

### `if-else`

-   `else`-statement is executed if `if`-condition equals `False`

``` PseudoPython
if condition:
   statement1
else:
   statement2
```

-   Notify indentation and colon

### Example

````python
%reset -f

# Source file: src/lesson_05_06/exampleif02.py

first=int(input('Give 1st number: '))
second=int(input('Give 2nd number: '))


sum=first+second
prod=first*second

if sum>=100:
    print(f'{first}+{second} is quite big number')
else:
    print(f'{first}+{second} is not that big')

if prod>=100:
    print(f'{first}*{second} is quite big number')
    print("It's nice to have large numbers")
else:
    print(f'{first}*{second} is not that big')
    print("It's nice to have small numbers")
````

````python
%reset -f

# Source file: src/lesson_05_06/exampleif03.py

ans=input('Do you like programming (y/n)? ')
ans=ans.upper() #Convert uppercase

if ans=='Y':
    print('Sounds good!')
else:
    print("That's too bad!")
    print('You should learn to like programming!')
````

### `if-elif-else`

-   The first `True` `if/elif`-statement is executed

-   If all condition equals to `False`, then `else`-block will be
    executed

-   `else` is optional

-   There can be any number of `elif`-blocks

``` PseudoPython
if condition1:
   statement1
elif condition2:
   statement2
elif condition3:
   statement3
...
else:
   statementN
```

### Example

````python
%reset -f

# Source file: src/lesson_05_06/exampleif04.py

grade=int(input('Give your grade (1-5): '))

if grade==1:
    print('Not too good!')
elif grade==2:
    print('You can do better!')
elif grade==3 or grade==4:
    print('Work harder next time!')
elif grade==5:
    print('Perfect!')
else:
    print('Not a valid grade!')
````

## `match-case`

### `match-case`

-   Python 3.10-

-   Compare value of a variable

-   To replace `if-elif-else`, in some cases

-   Easier to read

-   `case other:` can be replaced by `case _:`

````python
%reset -f

command = 'Hello, World!'
match command:
    case 'Hello, World!':
        print('Hello to you too!')
    case 'Goodbye, World!':
        print('See you later')
    case other:
        print('No match found')
````

### Example – `if-elif-else`

````python
%reset -f

# Source file: src/lesson_05_06/examplematch01.py

dow=int(input('Give day of week (1=Mon, ... 7=Sun): '))

if dow==1:
    dow_string='Mon'
elif dow==2:
    dow_string='Tue'
elif dow==3:
    dow_string='Wed'
elif dow==4:
    dow_string='Thu'
elif dow==5:
    dow_string='Fri'
elif dow==6:
    dow_string='Sat'
elif dow==7:
    dow_string='Sun'
else:
    dow_string='Unknown value'

print(f'You gave {dow_string}.')
````

### Example – `match-case`

````python
%reset -f

# Source file: src/lesson_05_06/examplematch02.py

dow=int(input('Give day of week (1=Mon, ... 7=Sun): '))

match dow:
    case 1:
        dow_string='Mon'
    case 2:
        dow_string='Tue'
    case 3:
        dow_string='Wed'
    case 4:
        dow_string='Thu'
    case 5:
        dow_string='Fri'
    case 6:
        dow_string='Sat'
    case 7:
        dow_string='Sun'
    case _:
        dow_string='Unknown value'

print(f'You gave {dow_string}.')
````

````python
%reset -f

# Source file: src/lesson_05_06/exampleconst03.py

from enum import Enum

class DayOfWeek(Enum):
    MON=1
    TUE=2
    WED=3
    THU=4
    FRI=5
    SAT=6
    SUN=7

dow=int(input('Give day of week (1=Mon, ... 7=Sun): '))

if dow<1 or dow>7:
    print('Not valid value!')
    exit()

match DayOfWeek(dow):
    case DayOfWeek.MON:
        dow_string='Mon'
    case DayOfWeek.TUE:
        dow_string='Tue'
    case DayOfWeek.WED:
        dow_string='Wed'
    case DayOfWeek.THU:
        dow_string='Thu'
    case DayOfWeek.FRI:
        dow_string='Fri'
    case DayOfWeek.SAT:
        dow_string='Sat'
    case DayOfWeek.SUN:
        dow_string='Sun'

print(f'You gave {dow_string}.')
````
