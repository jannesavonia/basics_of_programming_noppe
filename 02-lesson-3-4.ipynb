## Variables and data types

<hr style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### <span style="color:#1565c0;">Definition</span>

> A **variable** is an abstract storage location paired with an
> associated symbolic name, which contains some known or unknown
> quantity of information referred to as a value.

<hr style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

-   A named container for a particular type of data (like integer,
    float, string etc...)

-   Can eventually be associated with or identified by a memory address

-   Variable name is the usual way to reference the stored value

-   The value can be accessed or changed at any time

    > *In Python, not all languages!*

-   Initializing: `varname = value`

-   Updating: `varname = value`

### Data types

-   Text Type: `str`

    -   Values surrounded by `"` or `'`

    -   Multiline values `"""` or `'''`

-   Numeric Types: `int, float, complex`

    -   Decimal separator is `.`

    -   `3.0` is `float`

    -   0.1+2.1j is `complex`

-   Sequence Types: `list, tuple, range`

-   Mapping Type: `dict`

-   Set Types: `set, frozenset`

-   Boolean Type: `bool`

-   Binary Types: `bytes, bytearray, memoryview`

-   None Type: `NoneType`

### Assignment operator

-   `variable=expression`

-   Value of `variable` will be modified

-   Value of `expession` will replace existing value of `variable`

### Arithmetic operators

| `+`  | Addition                          |
|:----:|:----------------------------------|
| `-`  | Subtraction                       |
| `*`  | Multiplication                    |
| `**` | Exponentiation                    |
| `/`  | Division                          |
| `//` | Integer division (floor division) |
| `%`  | Modulus                           |

   
Result type of `a` $\diamond$ `b`:

| `type(a)` |      $\diamond$      | `type(b)` | result type |
|:---------:|:--------------------:|:---------:|:-----------:|
|   `int`   | `+, -, *, **, //, %` |   `int`   |    `int`    |
|   `int`   |         `/`          |   `int`   |   `float`   |
|  `float`  |         any          |   `int`   |   `float`   |
|   `int`   |         any          |  `float`  |   `float`   |
|  `float`  |         any          |  `float`  |   `float`   |

> *Result of `**` can be `complex`. For example `type((-1)**0.5)`
> returns `complex`*

.

### More arithmetic operators

| `a+=b`  | `a=a+b`  |
|:-------:|:---------|
| `a-=b`  | `a=a-b`  |
| `a*=b`  | `a=a*b`  |
| `a**=b` | `a=a**b` |
| `a/=b`  | `a=a/b`  |
| `a//=b` | `a=a//b` |
| `a%=b`  | `a=a%b`  |

-   Expressions surrounded by brackets will be evaluated first

### Example

````python
%reset -f

# Source file: src/lesson_03_04/exampleinput02.py

#Computes area of a regular polygon
#See https://www.wikihow.com/Find-the-Area-of-Regular-Polygons

#Import tan and pi from math library
from math import tan, pi

n=int(input('Number of sides: '))
S=float(input('Length of a side: '))

a=S/(2*tan(pi/n))
P=S*n
A=P*a/2

print(f'Area is {A:0.2f}')
````

````python
%reset -f

# Source file: src/lesson_03_04/examplearithmetic01.py

a=13
b=5

print(f'{a}+{b} =', a+b)
print(f'{a}-{b} =', a-b)
print(f'{a}*{b} =', a*b)
print(f'{a}/{b} =', a/b)
print(f'{a}//{b} =', a//b)
print(f'{a}%{b} =', a%b)
print(f'{a}**{b} =', a**b)
````
