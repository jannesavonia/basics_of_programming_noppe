## More on functions

### Named/Keyword arguments

-   So far, we have assigned function arguments by same order as they
    were in function definition

    -   This is the only way of writing arguments in many languages

-   In Python it is possible to utilize parameter names

### Example

```python
%reset -f

# Source file: src/lesson_15_16/examplemorefunc04.py

def print_sum(first, last):
    print(f'{first}+{last}={first+last}')
```

```python
#positional arguments
print_sum(1, 2)
```

```python
#one named argument
print_sum(1, last=2)
```

```python
#one named argument => Error
#Named arguments must be at end of the argument list
print_sum(first=1, 2)
```

```python
#two named arguments
print_sum(first=1, last=2)
```

```python
#two named arguments, see the output!
print_sum(last=1, first=2)
```

### Default parameter values

-   It is possible to give a default value for a parameter

-   If the value is not given, the default will be used

-   Default value can be given for none, one or many parameters

-   Default parameters must be the last parameters in parameter list

### Example

```python
%reset -f

# Source file: src/lesson_15_16/examplemorefunc01.py

from enum import Enum

class Language(Enum):
    FIN=1
    ENG=2
    SWE=3
```

```python
def say_hello(language=Language.ENG):
    msg={ Language.FIN: 'Hei!',
          Language.ENG: 'Hi!',
          Language.SWE: 'Hej!'
    }

    print(msg[language])
```

```python
say_hello(Language.FIN)
```

```python
say_hello(Language.ENG)
```

```python
say_hello(Language.SWE)
```

```python
say_hello()
```

```python
%reset -f

# Source file: src/lesson_15_16/examplemorefunc02.py

from enum import Enum

class Language(Enum):
    FIN=1
    ENG=2
    SWE=3

msg_hi={ Language.FIN: 'Hei!',
      Language.ENG: 'Hi!',
      Language.SWE: 'Hej!'
}

msg_bye={ Language.FIN: 'Moi!',
          Language.ENG: 'Bye!',
          Language.SWE: 'Hejsan!'
}

    
def show_message(language=Language.ENG, msg=msg_hi):
    print(msg[language])
```

```python
#Use dafault for msg
show_message(Language.FIN)
show_message(language=Language.SWE)
```

```python
#Use default for language and msg
show_message()
```

```python
#Use default for language
show_message(msg=msg_bye)
```

```python
#No dafaults
show_message(msg=msg_bye, language=Language.FIN)
show_message(Language.ENG, msg_bye)
```

### Arbitrary arguments – `*args`

-   If number of arguments is unknown

-   Use `*` before the parameter name in function definition

-   `tuple` of arguments will be received

```python
%reset -f

# Source file: src/lesson_15_16/examplemorefunc05.py

def my_func(*a):
    print(f'parameters = {a}')
```

```python
def my_func_multi(first, *a, last='foo'):
    print(f'first = {first}, arbitrary parameters = {a}, last={last}')
```

```python
my_func()
```

```python
my_func('one parameter')
```

```python
my_func('one', 2, 'three')
```

```python
print()

my_func_multi('one parameter')
```

```python
my_func_multi('one', 2, 'three')
```

```python
my_func_multi('one', 2, 'three', last='defined')
```

### Example

```python
%reset -f

# Source file: src/lesson_15_16/examplemorefunc06.py

def sum(*args):
    s=0
    for x in args:
        s+=x

    return s
```

```python
s=sum()
print(s, '?= 0, empty parameters')
```

```python
s=sum(1)
print(s, '?= 1, (1, )')
```

```python
s=sum(-1, -2)
print(s, '?= -3, (-1, -2)')
```

```python
s=sum(-11, 1.1)
print(s, '?= -9.9, (-11, -1.1)')
```

```python
s=sum(-11, 1.1, 9.9, 4.5, 1, 22.9)
print(s, '?= 28.4, (-11, 1.1, 9.9, 4.5, 1, 22.9)')
```

## Datetime

### Basics

-   Python contains module `datetime` to work with dates and times

-   `import datetime`

1.  Import `datetime` module

    -   Types (=classes) to represent dates and times

2.  Read current date and time from system clock

3.  Print the date and the time (current time)

```python
%reset -f

import datetime
```

```python
  now=datetime.datetime.now()
  print(now)
```

### Example – `date`

```python
%reset -f

# Source file: src/lesson_15_16/exampledatetime01.py

import datetime

today=datetime.date.today()
print(f'Today is {today}')
```

```python
arbitraryday=datetime.date(2023, 11, 30)
print(f'Arbitrary day is {arbitraryday}')
```

```python
diff=arbitraryday-today
print(f'There are {diff} from {today} to {arbitraryday}.')
```

```python
#Reformat earlier
print(f'There are {diff.days} days from {today} to {arbitraryday}.')
```

```python
tomorrow=today+datetime.timedelta(days=1)
print(f'Tomorrow will be {tomorrow}.')
```

```python
yesterday=today-datetime.timedelta(days=1)
print(f'Yeasterday was {yesterday}.')
#reformat earlier, see https://strftime.org/
print(f'Yeasterday was {yesterday.strftime("%A %d.%m.%Y")}.')
```

### Example – `datetime`

```python
%reset -f

# Source file: src/lesson_15_16/exampledatetime02.py

import datetime

now=datetime.datetime.now()
print(f'Now is {now}')
#reformat
print(f'Time is {now.strftime("%H:%M")}.')
print(f'Today is {now.strftime("%A %d.%m.%Y")}.')
```

```python
timein1hour=now+datetime.timedelta(hours=1)
print(f'In 1 hour it will be {timein1hour.strftime("%H:%M")}.')
```

```python
timein1week=now+datetime.timedelta(weeks=1)
print(f'In 1 week it will be {timein1week.strftime("%A %d.%m.%Y %H:%M")}.')
```

```python
diff=timein1week-timein1hour
print(f'There are {diff.seconds} seconds between {timein1hour} and {timein1week}')
```

### User input (date time strings)

-   There are number of different formats to represent date and time

    -   Finland 31.12.2023

    -   US 12-31-2023

    -   UK 31/12/2023

    -   Bonus: Different calendars

        -   Gregorian

        -   Hijri (Islamic)

        -   Chinese calendar

-   It is hard to write software that works with all variations

-   **Convert time string to `datetime` format as soon as possible**

-   If the string is asked from the user, make sure the user knows the
    format

### Example – User input

```python
%reset -f

# Source file: src/lesson_15_16/exampledatetime03.py

import datetime

input_str=input('Give date in dd.mm.yyyy format: ')
input_date=datetime.datetime.strptime(input_str, "%d.%m.%Y")
```

```python
#Convert to date
input_date=input_date.date()
```

```python
today=datetime.date.today()
diff=input_date-today

print(f'There are {diff.days} between today and given date.')
```
