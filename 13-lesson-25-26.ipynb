## Inheritance

### Example – Background

-   Suppose that we are designing veterinary clinic data system

-   The clinic takes care of different kind of animals, for example dogs
    and snakes. We will concentrate on these two types here

-   All pets share common data, for example

    -   owner’s name

    -   weight

    -   date of birth

-   Dog data

    -   breed

    -   height

    -   length

-   Snake data

    -   specie

    -   venomous

    -   length

    -   circumference

### Example – Data structure (pseudocode), version 1

``` PseudoPython
class Pet:
  owner
  weight
  date_of_birth
  
class Dog:
  breed
  height
  length
  owner
  weight
  date_of_birth

class Snake:
  specie
  venomous
  length
  circumference
  owner
  weight
  date_of_birth
```

### Example – Problems and solution

-   If `Pet` class is modified, we have to modify `Dog` and `Snake`

-   Logically `Dog` and `Snake` are `Pet`s

    -   There is no way to notice this in code

-   Solution:
    [*Inheritance*](https://en.wikipedia.org/wiki/Inheritance_(object-oriented_programming))

    -   `Pet` is a base class (parent class)

    -   Both, `Dog` and `Snake`, inherits `Pet`’s features

    -   Result will be similar to previous slide, except modifying `Pet`
        modifies automatically inherited classes

    -   Code tells us that `Dog` and `Snake` are `Pet`s

### Example – Data structure (pseudocode), version 2

``` PseudoPython
class Pet:
  owner
  weight
  date_of_birth
  
class Dog(Pet): #<- Contains Pet data
  breed
  height
  length

class snake(Pet): #<- Contains Pet data
  specie
  venomous
  length
  circumference
```

### OOP – Inheritance

-   Mechanism to create class based on the other class

    ``` PseudoPython
    class Parent:
      expressions
      
    class Child(Parent):
      more_expressions
    ```

-   `Child` contains `Parent`’s methods and data

-   `super()` to access parents methods or data

### Example

````python
%reset -f

# Source file: src/lesson_25_26/exampleclass09.py

class Pet:
    def __init__(self, owner, weight, dob):
        self.owner=owner
        self.weight=weight
        self.dob=dob

    def __str__(self):
        return f'Pet:{self.owner, self.weight, self.dob}'
````

````python
class Dog(Pet):
    def __init__(self, owner, weight, dob, breed, height, length):
        #super() - use parent class method
        super().__init__(owner, weight, dob)
        self.breed=breed
        self.height=height
        self.length=length
````

````python
class Snake(Pet):
    def __init__(self, owner, weight, dob, specie, venomous, length, circumference):
        super().__init__(owner, weight, dob)
        self.specie=specie
        self.venomous=venomous
        self.length=length
        self.circumference=circumference

    def __str__(self):
        return f'Snake:({super().__str__()}), {self.specie}'
````

````python
if __name__=='__main__':
    def print_object(name, o):
        print(f'{name} = {o}')
        print(f'  type: {type(o)}')
        print(f'  inherited: {o.__class__.__bases__}')
````

## OOP – Operator overloading

### Example

````python
%reset -f

# Source file: src/lesson_25_26/exampleclass10.py

class Fraction: #fraction m/n
    def __init__(self, m, n):
        self.m=m
        self.n=n

    def add(self, f):
        f=Fraction(self.m*f.n+self.n*f.m, self.n*f.n)
        return f

    def __str__(self):
        return f'({self.m}/{self.n})'

    def float(self):
        return self.m/self.n
````

````python
f1=Fraction(1, 4)
f2=Fraction(1, 3)
````

````python
f3=f1.add(f2)
````

````python
print(f'{f1}+{f2}={f3} approx {f3.float()}')
````

-   Is it possible to use notation `f3=f1+f2`?

### OOP – Operator overloading

-   Operator overloading is a method to define how objects act with `+`,
    `-`, `*`, `/` etc.

-   [Special function
    names](https://diveintopython3.net/special-method-names.html) are
    used

-   Improves readability

### Example

````python
%reset -f

# Source file: src/lesson_25_26/exampleclass11.py

class Fraction: #fraction m/n
    def __init__(self, m, n):
        self.m=m
        self.n=n

    def __add__(self, f):
        res=Fraction(self.m*f.n+self.n*f.m, self.n*f.n)
        return res

    def __str__(self):
        return f'({self.m}/{self.n})'

    def __float__(self):
        return self.m/self.n

    def __eq__(self, f):
        return self.m*f.n==f.m*self.n

    def __ne__(self, f):
        return not self==f
    
    def __gt__(self, f):
        return self.m*f.n>f.m*self.n

    def __lt__(self, f):
        return self.m*f.n<f.m*self.n
    
    def __ge__(self, f):
        return self.m*f.n>=f.m*self.n

    def __le__(self, f):
        return self.m*f.n<=f.m*self.n
````

````python
f1=Fraction(1, 4)
f2=Fraction(1, 3)
````

````python
f3=f1+f2
````

````python
print(f'{f1}+{f2}={f3} approx {float(f3)}')
````

````python
print(f'{f1}=={f2} = {f1==f2}')
print(f'{f1}!={f2} = {f1!=f2}')
print(f'{f1}>{f2} = {f1>f2}')
print(f'{f1}<{f2} = {f1<f2}')
print(f'{f1}>={f2} = {f1>=f2}')
print(f'{f1}<={f2} = {f1<=f2}')
````
