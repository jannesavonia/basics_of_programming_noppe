### Introduction to `for`-loop

````python
%reset -f

# Source file: src/lesson_07_08/exampleloop02.py

for i in range(10): #i=0...(10-1)
    print(i)
````

-   Block under `for` is repeated given number of times?

-   At first iteration a loop variable `i=0`

-   At second iteration `i=1`

-   …

-   At last iteration `i=9`

-   Note that $9=10-1$

### `range()`

-   `range(stop)`

    -   $0,\ 1, \dots\ \texttt{stop-1}$

-   `range(start, stop)`

    -   $\texttt{start},\ \texttt{start+1}, \dots\ \texttt{stop-1}$

-   `range(start, stop, step)`

    -   $\texttt{start},\ \texttt{start+step}, \dots\ \texttt{last}$

    -   $\texttt{last}<\texttt{stop}$

````python
%reset -f

# Source file: src/lesson_07_08/exampleloop05.py

N=10

#Print header row
print('\t', end='')
for i in range(1, N+1):
    print(i, end='\t')
print()

#Print horixontal bar
for i in range(N+1):
    print(7*'-', end='\t')
print()

for i in range(1, N+1):
    #Print 1st column
    print(i, end='     \t|')

    #print result
    for j in range(1, N+1):
        print(i*j, end='\t')
    print()
````

### `continue` and `break`

-   `break` terminates the loop

-   `continue` skips current iteration, but loop goes to next iteration

````python
%reset -f

# Source file: src/lesson_07_08/exampleloop06.py

#Print primes below 100

N=100
for candidate in range(2, 100):
    is_prime=True
    for test in range(2, candidate-1):
        if candidate%test==0:
            is_prime=False
            break #Found divisor, no point to test more numbers
    if is_prime:
        print(candidate)
````

### Example

````python
%reset -f

# Source file: src/lesson_07_08/exampleloop07.py

#Prints positive numbers 0..19, except
#those divisible by 3 or 7

N=20

for i in range(N):
    if i%3==0 or i%7==0:
        continue
    print(i)
````

## `while`

### `while`

-   `while`-loop is repeated as long as end condition is `True`

-   More flexible than `for`-loop

``` PseudoPython
while condition:
   statement
```

-   `break` and `continue` – similar to `for`

### Example

````python
%reset -f

# Source file: src/lesson_07_08/exampleloop08.py

money=100.0

while money>=0:
    print('Money: '+str(money))
    deposit=float(input('Give deposit (negative=draw out): '))
    money+=deposit

print('Out of money!')
````

### Example

-   `while`-loop can be used to execute the block at least once

````python
%reset -f

# Source file: src/lesson_07_08/exampleloop09.py

while True:
    n=int(input('Give integer 1..3: '))
    if n>=1 and n<=3:
        break

    print('Try again!')

print(f'You gave {n}')
````

### Example – Nested loops

````python
%reset -f

# Source file: src/lesson_07_08/exampleloop10.py

while True:
    print('Give 0 to terminate!')
    n=int(input('What multiplication table you like to see? '))

    if n==0:
        break
    
    for i in range(1, 11):
        print(f'{n}*{i} = {n*i}')

    print(32*'-')
````
