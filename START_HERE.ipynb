# Basics of Programming

Run code below to copy notebooks to `my-work/bop_notebooks`. To run the code, click the code and press Ctrl+Enter.

Existing notebooks in `my-work/bop_notebooks` will not be overwritten.

After this, start [my-work/bop_notebooks/00-index.ipynb](../my-work/bop_notebooks/00-index.ipynb).

```python
from pathlib import Path
import shutil

HOME = Path.home()
SOURCE = HOME / "basics_of_programming_noppe"
MY_WORK = HOME / "my-work"
DESTINATION = MY_WORK / "bop_notebooks"

# Files that must not be copied to the student's workspace.
EXCLUDED_FILES = {
    "START_HERE.ipynb",
}


def copy_course_material():
    print("Basics of Programming")
    print("Updating course material...")
    print()

    if not SOURCE.exists():
        print("ERROR: Course material directory was not found:")
        print(f"  {SOURCE}")
        return

    if not MY_WORK.exists():
        print("ERROR: my-work directory was not found:")
        print(f"  {MY_WORK}")
        print()
        print("Please contact the course teacher.")
        return

    DESTINATION.mkdir(parents=True, exist_ok=True)

    copied = []
    existing = []
    errors = []

    # Copy all course files and preserve the directory structure.
    for source_file in sorted(SOURCE.rglob("*")):

        # Directories themselves do not need to be copied.
        # They are created when their files are copied.
        if source_file.is_dir():
            continue

        # Do not copy the utility notebook itself.
        if source_file.name in EXCLUDED_FILES:
            continue

        relative_path = source_file.relative_to(SOURCE)
        destination_file = DESTINATION / relative_path

        # Never overwrite the student's existing files.
        if destination_file.exists():
            existing.append(relative_path)
            continue

        try:
            destination_file.parent.mkdir(
                parents=True,
                exist_ok=True
            )

            shutil.copy2(
                source_file,
                destination_file
            )

            copied.append(relative_path)

        except Exception as error:
            errors.append(
                (relative_path, error)
            )

    # --------------------------------------------------------
    # Report
    # --------------------------------------------------------

    if copied:
        print(f"Copied {len(copied)} new file(s):")

        for filename in copied:
            print(f"  + {filename}")
    else:
        print("No new files to copy.")

    print()

    if existing:
        print(
            f"Kept {len(existing)} existing file(s) unchanged."
        )

    if errors:
        print()
        print("Some files could not be copied:")

        for filename, error in errors:
            print(f"  ! {filename}: {error}")

    print()
    print("Course material is available in:")
    print(f"  {DESTINATION}")
    print()

    if not errors:
        print("Course material is ready.")
        print()
        print("Open the course index:")
        print(f"  {DESTINATION / '00-index.ipynb'}")


copy_course_material()
```
