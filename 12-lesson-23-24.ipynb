### Solution – getter and setter

-   In earlier example, accessing `team` variable corrupted internal
    state of the class

    -   Rule of thumb: Information shall be stored only once

    -   At the example `team` variable can be computed from
        `SokkerPlayer.teams`

-   We can solve the problem by replacing `team` variable by
    `set_team(team)` function

-   In addition, we need to create `get_team()` function

```python
%reset -f

# Source file: src/lesson_23_24/exampleclass05.py

class SokkerPlayer:
    teams={}

    def __init__(self, name, team, age):
        self.name=name
        self.age=age
        #set_team utilizes name! Therefore,
        #it must be located after self.name=name
        self.set_team(team)

    def set_team(self, team):
        print(f'{self.name}.set_team({team})')
        SokkerPlayer.teams[self.name]=team

    def get_team(self):
        return SokkerPlayer.teams[self.name]
        
    def teammates(self):
        return [n for n, t in SokkerPlayer.teams.items() if t==self.get_team()]
```

```python
matti=SokkerPlayer('Matti', 'KUPS', 14)
kaisa=SokkerPlayer('Kaisa', 'Kuopion Elo', 15)
kalle=SokkerPlayer('Kalle', 'Kuopion Elo', 12)

print('Kalle\'s teammates are ', kalle.teammates())
```

```python
#It is possible to access class/instance variables
print(SokkerPlayer.teams)
print('Matti\'s team is', matti.get_team())
matti.set_team('Kuopion Elo')
print('Matti\'s team is', matti.get_team())
print('Kalle\'s teammates are ', kalle.teammates())
```

### Getter and setter

-   Preferred solution for the earlier problem is that we can access
    `matti.team` variable safe

-   The solution is to utilize getter and setter

-   In Python, [getter and
    setter](https://realpython.com/python-getter-setter/) can make work
    by the variable name

    -   `matti.team=’Kuopion Elo’` calls automatically
        `matti.set_team(’Kuopion Elo’)`

    -   `print(matti.team)` calls automatically `matti.get_team()`

-   Setters/getters can be utilized for example:

    -   Eliminate duplicate information

    -   Check validity of data – For example, check that variable value
        is $-3\dots 8$

    -   Convert data to preferred format – For example, convert string
        to uppercase

### Example – with functions

```python
%reset -f

# Source file: src/lesson_23_24/exampleclass06.py

class Student:
    def __init__(self, firstname, lastname, weight):
        self.firstname=firstname
        self.lastname=lastname
        self.set_weight(weight)

    def set_name(self, fullname):
        firstname, lastname=fullname.split()
        self.firstname=firstname
        self.lastname=lastname
        
    def get_name(self):
        return self.firstname+' '+self.lastname

    def set_weight(self, weight):
        if weight<0 or weight>500:
            raise ValueError
        self.weight=weight
```

```python
matti=Student('Matti', 'Mainio', 85.5)
jussi=Student('Jussi', 'Juonio', 78.8)

print(matti.get_name())
print(jussi.get_name())
```

```python
matti.set_name('Matt Superb')
print(matti.get_name())
```

```python
try:
    matti.set_weight(-10.2)
except:
    print('Got an exception!')
```

```python
matti.set_weight(87.7)
print('Matt\'s weight is',matti.weight)
```

### Example – with getter/setter

```python
%reset -f

# Source file: src/lesson_23_24/exampleclass07.py

class Student:
    def __init__(self, firstname, lastname, weight):
        print('  constructor')
        self.firstname=firstname
        self.lastname=lastname
        self.weight=weight

    @property
    def name(self):
        print('  name getter')
        return self.firstname+' '+self.lastname

    @name.setter
    def name(self, fullname):
        print('  name setter')
        firstname, lastname=fullname.split()
        self.firstname=firstname
        self.lastname=lastname

    @property
    def weight(self):
        print('  weight getter')
        #Use protected variable _weight to store weight
        return self._weight

    @weight.setter
    def weight(self, weight):
        print('  weight setter')
        if weight<0 or weight>500:
            raise ValueError
        self._weight=weight
```

```python
if __name__=='__main__':
    matti=Student('Matti', 'Mainio', 85.5)
    jussi=Student('Jussi', 'Juonio', 78.8)
```
